```python
from transformers import pipeline, AutoTokenizer, AutoModel
import torch
```

```python
from transformers import pipeline

mask_filler = pipeline("fill-mask")

input_sentence = "Hanoi is the <mask> of Vietnam."

predictions = mask_filler(input_sentence, top_k=5)

print("Câu gốc:", input_sentence)
for pred in predictions:
    print(f"Dự đoán: '{pred['token_str']}' với độ tin cậy: {pred['score']:.4f}")
    print(" ->", pred['sequence'])

```

```python
# Câu hỏi 1: Mô hình dự đoán 'capital' hay không
# Câu hỏi 2: Vì sao BERT phù hợp cho MLM?

answer_1 = """
1. CÓ! Mô hình dự đoán đúng từ ‘capital’ Và không chỉ đúng, mà còn đúng với độ tin cậy rất cao (0.9341).
2. BERT phù hợp với Masked Language Modeling vì nó là mô hình Encoder-only,
   có khả năng nhìn hai chiều (bidirectional), nên hiểu được ngữ cảnh cả trước và sau token bị mask.
"""
print(answer_1)
```

```python
# 1. Load pipeline text-generation (GPT-2 mặc định)
generator = pipeline("text-generation")

# 2. Prompt ban đầu
prompt = "The best thing about learning NLP is"

# 3. Sinh văn bản tiếp theo
generated_texts = generator(prompt, max_length=50, num_return_sequences=1)

# 4. In kết quả
print("Câu mồi:", prompt)
for text in generated_texts:
    print("\nVăn bản được sinh ra:")
    print(text["generated_text"])
```

```python
answer_2 = """
1. CÓ – kết quả GPT-2 sinh ra hoàn toàn hợp lý theo hành vi của mô hình GPT-2.
2. GPT (Decoder-only) phù hợp với Next Token Prediction vì nó chỉ nhìn một chiều
   (unidirectional) — dự đoán token tiếp theo dựa trên các token trước đó.
"""
print(answer_2)
```

```python
# 1. Chọn mô hình
model_name = "bert-base-uncased"
tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModel.from_pretrained(model_name)

# 2. Input phrases
sentences = ["This is a sample sentence."]

# 3. Tokenization
inputs = tokenizer(sentences, padding=True, truncation=True, return_tensors='pt')

# 4. Lấy hidden states
with torch.no_grad():
    outputs = model(**inputs)

last_hidden_state = outputs.last_hidden_state  # (batch, seq_len, hidden)
attention_mask = inputs["attention_mask"]

# 5. Mean pooling (bỏ padding)
mask_expanded = attention_mask.unsqueeze(-1).expand(last_hidden_state.size()).float()
sum_embeddings = torch.sum(last_hidden_state * mask_expanded, dim=1)
sum_mask = torch.clamp(mask_expanded.sum(1), min=1e-9)

sentence_embedding = sum_embeddings / sum_mask

# 6. In vector kết quả
print("Embedding của câu:")
print(sentence_embedding)
print("\nKích thước vector:", sentence_embedding.shape)
```

```python
answer_3 = """
1. Vector có kích thước (1, 768). Đây là hidden_size = 768 của BERT-base.
2. attention_mask được dùng trong Mean Pooling để không tính trung bình qua các token PAD.
"""
print(answer_3)
```
