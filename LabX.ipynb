```python
import numpy as np
from IPython.display import Audio

# Tạo âm thanh sóng sin mô phỏng DL TTS
def fake_tts_level2(text, duration=2.0, sample_rate=22050):
    t = np.linspace(0, duration, int(sample_rate * duration))
    base_freq = 180

    # Tạo dao động dựa trên độ dài text (mô phỏng phần acoustic model)
    freq = base_freq + (len(text) % 20) * 10

    waveform = 0.3 * np.sin(2 * np.pi * freq * t)
    return waveform, sample_rate

audio, sr = fake_tts_level2("This is simulated level 2 TTS using a fake deep learning pipeline.")
Audio(audio, rate=sr)


```
